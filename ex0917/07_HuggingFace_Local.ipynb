```python
from dotenv import load_dotenv
from langchain_teddynote import logging

# .env 정보 불러오기
load_dotenv()

# LangSmith 추적
logging.langsmith("CH04-Models")
```

```python
import os

# Hugging Face 모델을 ex0917/cache 폴더에 저장
os.environ["HF_HOME"] = "./cache/"
```

```python
from langchain_huggingface import HuggingFacePipeline

# 작은 모델을 PC에 다운로드해서 직접 실행
llm = HuggingFacePipeline.from_model_id(
    model_id="google/flan-t5-small",
    task="text2text-generation",
    device=-1,  # CPU 사용
    pipeline_kwargs={
        "max_new_tokens": 100,
        "do_sample": False,
    },
)
```

```python
response = llm.invoke(
    "What is the capital of South Korea?"
)

print(response)
```

```python
from langchain_core.prompts import PromptTemplate

template = """
Summarize the following text in simple bullet points.

TEXT:
{text}

Key Points:
"""

# 프롬프트 생성
prompt = PromptTemplate.from_template(template)

# 프롬프트 → 로컬 모델
chain = prompt | llm
```

```python
text = """
Large Language Models are AI models trained on large amounts of text.
They can answer questions, summarize documents, write code, and generate text.
However, their answers can sometimes be inaccurate.
"""

response = chain.invoke(
    {"text": text}
)

print(response)
```
