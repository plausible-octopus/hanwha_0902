# EnumOutputParser

`EnumOutputParser`는 LLM의 출력 결과를 미리 정해둔 `Enum` 값 중 하나로 변환해 주는 출력 파서입니다.

## 핵심 특징

- **Enum 값으로 변환**: 문자열 결과를 미리 정의한 Enum 값으로 바꿔줍니다.
- **결과 제한**: 지정해둔 값 중 하나만 결과로 나오도록 할 수 있습니다.
- **출력 정리**: 공백이나 줄바꿈이 포함되어도 적절하게 처리합니다.

## 언제 사용하나?

답변을 여러 선택지 중 하나로 제한하고 싶을 때 유용합니다.

예를 들어 색상을 `빨간색 / 초록색 / 파란색` 중 하나로만 출력하도록 만들 수 있습니다.

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
# LangSmith 추적 설정
# !pip install langchain-teddynote
from langchain_teddynote import logging

# 프로젝트 이름 설정
logging.langsmith("CH03-OutputParser")
```

```python
from langchain_classic.output_parsers.enum import EnumOutputParser
```

```python
from enum import Enum


class Colors(Enum):
    RED = "빨간색"
    GREEN = "초록색"
    BLUE = "파란색"
```

앞에서 만든 `Colors` Enum을 기준으로 `EnumOutputParser`를 생성합니다.

```python
# Colors 값을 사용하는 EnumOutputParser 생성
parser = EnumOutputParser(enum=Colors)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI

# 프롬프트 템플릿 생성
prompt = PromptTemplate.from_template(
    """다음의 물체는 어떤 색깔인가요?

Object: {object}

Instructions: {instructions}"""
).partial(instructions=parser.get_format_instructions())

# Prompt → LLM → OutputParser 순서로 연결
chain = prompt | ChatOpenAI() | parser
```

`chain.invoke()`를 사용해서 **하늘의 색상**을 질문합니다.

```python
# 체인 실행
response = chain.invoke({"object": "하늘"})
print(response)
```

반환된 결과의 타입과 실제 값을 확인합니다.

```python
# 결과 타입 확인
type(response)
```

```python
# Enum에 저장된 실제 값 확인
response.value
```
